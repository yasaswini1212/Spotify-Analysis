```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Selecting numerical columns for PCA
numerical_data = df.select_dtypes(include=['float64'])

# Standardizing the data
scaler = StandardScaler()
scaled_data = scaler.fit_transform(numerical_data)

# Applying PCA
pca = PCA()
pca_components = pca.fit_transform(scaled_data)

# Explained variance ratio
explained_variance = pca.explained_variance_ratio_

# Plotting explained variance
plt.figure(figsize=(10, 6))
sns.barplot(x=list(range(1, len(explained_variance) + 1)), y=explained_variance, palette='coolwarm')
plt.title('Explained Variance Ratio by Principal Component')
plt.xlabel('Principal Component')
plt.ylabel('Variance Explained')
plt.show()

# Cumulative explained variance
cumulative_variance = explained_variance.cumsum()

# Plotting cumulative variance
plt.figure(figsize=(10, 6))
sns.lineplot(x=list(range(1, len(cumulative_variance) + 1)), y=cumulative_variance, marker='o', color='blue')
plt.axhline(y=0.95, color='red', linestyle='--', label='95% Variance Threshold')
plt.title('Cumulative Explained Variance')
plt.xlabel('Number of Principal Components')
plt.ylabel('Cumulative Variance Explained')
plt.legend()
plt.show()

# Determining number of components to retain
components_to_retain = (cumulative_variance >= 0.95).argmax() + 1

components_to_retain
```

```python
# Loudness category
df['loudness_category'] = pd.cut(
    df['loudness'],
    bins=[-float('inf'), -40, -20, float('inf')],
    labels=['Low', 'Medium', 'High']
)

# Danceability category
df['danceability_category'] = pd.qcut(
    df['danceability'],
    q=3,
    labels=['Low', 'Medium', 'High']
)
```

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
import numpy as np
import matplotlib.pyplot as plt
# Selecting numerical features for transformations
numerical_columns = ['danceability', 'energy', 'loudness', 'valence', 'tempo']

# Normalization using Min-Max Scaling
normalized_data = pd.DataFrame(
    MinMaxScaler().fit_transform(df[numerical_columns]),
    columns=[f"{col}_normalized" for col in numerical_columns]
)

# Standardization using Z-Score Scaling
standardized_data = pd.DataFrame(
    StandardScaler().fit_transform(df[numerical_columns]),
    columns=[f"{col}_standardized" for col in numerical_columns]
)

# Log Transformation (adding a small constant to avoid log(0))
log_transformed_data = pd.DataFrame(
    np.log1p(df[numerical_columns].clip(lower=0.01)),
    columns=[f"{col}_log" for col in numerical_columns]
)

# Combining all transformed datasets
transformed_data = pd.concat([normalized_data, standardized_data, log_transformed_data], axis=1)

# Plotting distributions of original and transformed data for comparison
plt.figure(figsize=(20, 15))
for i, col in enumerate(numerical_columns):
    plt.subplot(5, 3, i * 3 + 1)
    sns.histplot(df[col], kde=True, color='blue')
    plt.title(f'Original {col}')

    plt.subplot(5, 3, i * 3 + 2)
    sns.histplot(normalized_data[f"{col}_normalized"], kde=True, color='green')
    plt.title(f'Normalized {col}')

    plt.subplot(5, 3, i * 3 + 3)
    sns.histplot(log_transformed_data[f"{col}_log"], kde=True, color='orange')
    plt.title(f'Log Transformed {col}')

plt.tight_layout()
plt.show()

# Displaying a snippet of the transformed data
transformed_data.head()
```

Hyper Parameter Tuning

```python
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.neighbors import KNeighborsClassifier
import pandas as pd
data = df.drop(columns=['track_name', 'artist', 'album'])
label_encoder = LabelEncoder()
data['genre'] = label_encoder.fit_transform(data['genre'])
X = data.drop(columns=['genre'])
y = data['genre']
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# KNN and Grid Search
knn = KNeighborsClassifier()
param_grid = {'n_neighbors': [3, 5, 7, 9], 'weights': ['uniform', 'distance'], 'p': [2]}
grid_search = GridSearchCV(knn, param_grid, cv=5, scoring='accuracy', verbose=1)
grid_search.fit(X_train, y_train)

print("Best Parameters:", grid_search.best_params_)
print("Best Cross-Validated Accuracy:", grid_search.best_score_)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Drop irrelevant columns
spotify_data_processed = df.drop(columns=['track_name', 'artist', 'album'])

# Encode the 'genre' column
label_encoder = LabelEncoder()
spotify_data_processed['genre'] = label_encoder.fit_transform(spotify_data_processed['genre'])

# Separate features and target
X = spotify_data_processed.drop(columns=['genre'])
y = spotify_data_processed['genre']

# Normalize the numerical features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

X_train.shape, X_test.shape, y_train.shape, y_test.shape
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Convert results to a DataFrame
results = pd.DataFrame(grid_search.cv_results_)
```

```python
# Filter results for 'distance' weights
heatmap_data = results[results['param_weights'] == 'distance'].pivot(
    index='param_n_neighbors', columns='param_p', values='mean_test_score'
)

# Plot heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(heatmap_data, annot=True, cmap='coolwarm', fmt='.3f', cbar_kws={'label': 'Accuracy'})
plt.title('Heatmap of Accuracy for Distance Weights')
plt.xlabel('p (Distance Metric)')
plt.ylabel('n_neighbors')
plt.show()
```

```python
for weight in ['uniform', 'distance']:
    subset = results[results['param_weights'] == weight]
    plt.plot(subset['param_n_neighbors'], subset['mean_test_score'], label=f'Weight: {weight}')

plt.title('Accuracy vs. n_neighbors')
plt.xlabel('Number of Neighbors')
plt.ylabel('Accuracy')
plt.legend()
plt.show()
```

```python
weight_data = results.groupby('param_weights')['mean_test_score'].mean()
weight_data.plot(kind='bar', color=['blue', 'orange'])
plt.title('Accuracy by Weight Function')
plt.ylabel('Mean Accuracy')
plt.show()
```
